# SweetViz trên Marketing Campaign

## Mục tiêu
- Tìm hiểu EDA tự động bằng SweetViz.
- Tạo báo cáo HTML về phân phối biến, giá trị thiếu, giá trị duy nhất, thống kê mô tả và liên hệ giữa biến.
- Thử phân tích biến mục tiêu `Response` nếu cột này tồn tại.

## Dữ liệu
Tải `marketing_campaign.csv` từ [Customer Personality Analysis trên Kaggle](https://www.kaggle.com/datasets/imakash3011/customer-personality-analysis) và đặt trong thư mục `data/`. Bộ dữ liệu thường dùng dấu phân cách tab; notebook tự thử đọc bằng tab trước.

```python
from pathlib import Path
import pandas as pd
DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)
files = list(DATA_DIR.glob("*.csv"))
DATA_FILE = next((p for p in files if "marketing" in p.name.lower()), files[0] if files else None)
if DATA_FILE is None:
    raise FileNotFoundError("Hãy đặt marketing_campaign.csv trong thư mục data/.")
try:
    df = pd.read_csv(DATA_FILE, sep="\t")
    if df.shape[1] == 1:
        df = pd.read_csv(DATA_FILE)
except Exception:
    df = pd.read_csv(DATA_FILE)
print("co" + str(df.shape[0]) + " dong va " + str(df.shape[1]) + " cot")
display(df.head())
display(df.dtypes.to_frame("dtype"))
```

## 1. Cài đặt và nạp SweetViz

```python
import sweetviz as sv
print("SweetViz đã được import.")
```

## 2. Kiểm tra chất lượng dữ liệu trước khi diễn giải báo cáo

```python
display(pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing": df.isna().sum(),
    "missing_percent": (df.isna().mean()*100).round(2),
    "unique": df.nunique(dropna=True)
}).sort_values("missing", ascending=False))
print("Dòng trùng lặp:", df.duplicated().sum())
```

## 3. Tạo báo cáo EDA tự động

```python
report = sv.analyze(df)
report.show_html(filepath="SweetViz_Marketing_Report.html", open_browser=False)
print("Đã lưu báo cáo: SweetViz_Marketing_Report.html")
```

## 4. Phân tích biến mục tiêu (nếu có)
Trong bộ dữ liệu Customer Personality Analysis, `Response` thường biểu thị khách hàng có phản hồi với chiến dịch marketing hay không. Phân tích target giúp xem các đặc trưng khác nhau như thế nào theo biến mục tiêu; đây là mô tả, không chứng minh quan hệ nhân quả.

```python
if "Response" in df.columns:
    target_report = sv.analyze(df, target_feat="Response")
    target_report.show_html(filepath="SweetViz_Marketing_Response_Report.html", open_browser=False)
    print("Đã lưu báo cáo có target Response.")
else:
    print("Không có cột Response; bỏ qua phân tích target.")
```

## 5. So sánh hai nhóm khách hàng
Ví dụ so sánh khách hàng phản hồi và không phản hồi chiến dịch, nếu `Response` có trong dữ liệu.

```python
if "Response" in df.columns and df["Response"].nunique(dropna=True) == 2:
    groups = df["Response"].dropna().unique()
    # compare_intra nhận điều kiện Boolean; báo cáo tên hai nhóm theo điều kiện True/False.
    mask = df["Response"] == groups[0]
    compare_report = sv.compare_intra(df, mask, [f"Response={groups[0]}", f"Response!={groups[0]}"])
    compare_report.show_html(filepath="SweetViz_Response_Group_Comparison.html", open_browser=False)
    print("Đã lưu báo cáo so sánh nhóm.")
else:
    print("Không đủ điều kiện để so sánh hai nhóm Response.")
```

## 6. Cách đọc và kết luận
- **Overview/Summary:** số dòng, biến, kiểu dữ liệu và các cảnh báo chất lượng.
- **Variables:** phân phối, thống kê tóm tắt, giá trị thiếu/duy nhất và các giá trị phổ biến.
- **Associations/Correlations:** mối liên hệ thống kê phù hợp với kiểu biến; tương quan không đồng nghĩa nhân quả.
- **Target analysis:** nếu dùng `Response`, quan sát đặc trưng khác biệt giữa các nhóm.
- Mở file HTML trong thư mục notebook để xem tương tác và ghi lại ít nhất 3 phát hiện có số liệu cụ thể.
