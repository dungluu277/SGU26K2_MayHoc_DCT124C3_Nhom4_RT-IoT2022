#  AutoViz trên Marketing Campaign

## Mục tiêu
- Tìm hiểu AutoViz, thư viện tự động sinh biểu đồ phục vụ EDA.
- Khảo sát phân phối biến số, biến phân loại và mối quan hệ giữa các biến.
- Lưu biểu đồ để đưa vào báo cáo và so sánh trải nghiệm với SweetViz.

## Dữ liệu
Tải `marketing_campaign.csv` từ [Customer Personality Analysis trên Kaggle](https://www.kaggle.com/datasets/imakash3011/customer-personality-analysis) và đặt trong thư mục `data/`. File phổ biến dùng dấu tab làm dấu phân cách.

```python
from pathlib import Path
import pandas as pd
DATA_DIR = Path("data")
PLOTS_DIR = Path("AutoViz_Plots")
PLOTS_DIR.mkdir(parents=True, exist_ok=True)
DATA_DIR.mkdir(exist_ok=True)
files = list(DATA_DIR.glob("*.csv"))
DATA_FILE = next((p for p in files if "marketing" in p.name.lower()), files[0] if files else None)
if DATA_FILE is None:
    raise FileNotFoundError("Hãy đặt marketing_campaign.csv trong thư mục data/.")
try:
    df = pd.read_csv(DATA_FILE, sep="\t")
    if df.shape[1] == 1:
        df = pd.read_csv(DATA_FILE)
except Exception:
    df = pd.read_csv(DATA_FILE)
print("File:", DATA_FILE, "| Kích thước:", df.shape)
display(df.head())
display(df.describe(include="all").T)
```

## 2. Sinh biểu đồ tự động

```python
from autoviz import AutoViz_Class
AV = AutoViz_Class()
```

AutoViz hỗ trợ đầu vào là đường dẫn file hoặc DataFrame. Để tránh phụ thuộc vào tên file cụ thể, bài thực hành truyền DataFrame đã nạp. `verbose=2` lưu biểu đồ PNG vào thư mục `AutoViz_Plots/`; số dòng/cột phân tích được giới hạn để giảm thời gian chạy.

```python
# AutoViz thường hoạt động tốt với dữ liệu có số lượng cột vừa phải.
# depVar để trống vì mục tiêu chính là EDA tổng quát, không huấn luyện mô hình.
autoviz_result = AV.AutoViz(
    filename="",
    sep="\t",
    depVar="",
    dfte=df,
    header=0,
    verbose=2,
    lowess=False,
    chart_format="png",
    max_rows_analyzed=150000,
    max_cols_analyzed=30,
    save_plot_dir=str(PLOTS_DIR)
)
print("AutoViz đã chạy. Nếu phiên bản cài đặt không nhận DataFrame qua dfte, xem ô thay thế bên dưới.")
```

## 3. Phương án thay thế nếu API phiên bản cài đặt khác

```python
# Dùng ô này nếu lời gọi phía trên báo lỗi do khác biệt API giữa các phiên bản.
# Lưu DataFrame ra CSV tạm để AutoViz đọc theo đường dẫn.
temp_file = DATA_DIR / "marketing_campaign_for_autoviz.csv"
df.to_csv(temp_file, index=False)
autoviz_result = AV.AutoViz(
    str(temp_file),
    sep=",",
    depVar="",
    dfte=None,
    header=0,
    verbose=2,
    lowess=False,
    chart_format="png",
    max_rows_analyzed=150000,
    max_cols_analyzed=30,
    save_plot_dir=str(PLOTS_DIR)
)
```
