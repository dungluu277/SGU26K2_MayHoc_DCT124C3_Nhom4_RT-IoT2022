# 1. Import thư viện và đọc dữ liệu

```python
from pathlib import Path
import inspect
import json
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    FunctionTransformer,
    StandardScaler,
    OneHotEncoder
)
from sklearn.pipeline import Pipeline
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import StratifiedKFold

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    precision_recall_fscore_support,
    confusion_matrix
)

# ========================================
# 1. CẤU HÌNH THỰC NGHIỆM
# ========================================

EXPERIMENT_ID = "E12"
COMPARE_ID = "E10"

TARGET = "Attack_type"
RANDOM_STATE = 42
N_SPLITS = 3

# ========================================
# 2. TÌM THƯ MỤC GỐC IoT_final
# ========================================

CURRENT = Path.cwd().resolve()

ROOT = next(
    (
        p for p in [CURRENT, *CURRENT.parents]
        if (p / "project").is_dir()
        and (p / "result").is_dir()
    ),
    None
)

if ROOT is None:
    raise FileNotFoundError(
        "Không tìm thấy IoT_final. "
        "Hãy mở VS Code tại thư mục dự án."
    )

# ========================================
# 3. ĐƯỜNG DẪN TRAIN VÀ VALIDATION
# ========================================

DATA_DIR = (
    ROOT / "result"
    / "preprocessing"
    / "C02_Data_Splitting_ThanhDai"
    / "data"
    / "processed"
)

TRAIN_PATH = DATA_DIR / "train.csv"
VALID_PATH = DATA_DIR / "validation.csv"

# ========================================
# 4. THƯ MỤC KẾT QUẢ E12
# ========================================

RESULT_DIR = (
    ROOT / "result"
    / "exps"
    / "E12_MLP_SignedLog"
)

RESULT_DIR.mkdir(parents=True, exist_ok=True)

# ========================================
# 5. ĐỌC DỮ LIỆU
# ========================================

if not TRAIN_PATH.is_file():
    raise FileNotFoundError(TRAIN_PATH)

if not VALID_PATH.is_file():
    raise FileNotFoundError(VALID_PATH)

train_df = pd.read_csv(TRAIN_PATH)
valid_df = pd.read_csv(VALID_PATH)

print("=== THỰC NGHIỆM E12 ===")
print("MLP + Signed-log + StandardScaler")

print("\n=== KÍCH THƯỚC DỮ LIỆU ===")
print("Train:", train_df.shape)
print("Validation:", valid_df.shape)

print("\n=== THƯ MỤC LƯU KẾT QUẢ ===")
print(RESULT_DIR)

print("\n=== 5 DÒNG TRAIN ĐẦU TIÊN ===")
display(train_df.head())
```

# 2. Xây dựng Pipeline Signed_Lod + StandardScaler + MLP

```python
# ========================================
# 1. TÁCH ĐẶC TRƯNG VÀ NHÃN
# ========================================

DROP_COLS = [
    TARGET,
    "Unnamed: 0",
    "group_id"
]

CATEGORICAL_FEATURES = [
    "proto",
    "service"
]

if (
    TARGET not in train_df.columns
    or TARGET not in valid_df.columns
):
    raise ValueError(
        f"Thiếu cột nhãn {TARGET}"
    )

if (
    train_df[TARGET].isna().any()
    or valid_df[TARGET].isna().any()
):
    raise ValueError("Nhãn chứa giá trị thiếu")

X = train_df.drop(
    columns=DROP_COLS,
    errors="ignore"
).copy()

y = train_df[TARGET].astype(str).copy()

X_valid = valid_df.drop(
    columns=DROP_COLS,
    errors="ignore"
).copy()

y_valid = valid_df[TARGET].astype(str).copy()

# Đồng bộ cột Train và Validation
if set(X.columns) != set(X_valid.columns):
    raise ValueError(
        "Train và Validation có tập đặc trưng khác nhau"
    )

X_valid = X_valid[X.columns]

# ========================================
# 2. PHÂN LOẠI ĐẶC TRƯNG
# ========================================

missing_categorical = [
    c for c in CATEGORICAL_FEATURES
    if c not in X.columns
]

if missing_categorical:
    raise ValueError(
        f"Thiếu cột categorical: {missing_categorical}"
    )

NUMERIC_FEATURES = X.select_dtypes(
    include=np.number
).columns.tolist()

unknown_features = sorted(
    set(X.columns)
    - set(NUMERIC_FEATURES)
    - set(CATEGORICAL_FEATURES)
)

if unknown_features:
    raise ValueError(
        f"Các cột chưa xử lý: {unknown_features}"
    )

# ========================================
# 3. KIỂM TRA DỮ LIỆU
# ========================================

if X.isna().any().any() or X_valid.isna().any().any():
    raise ValueError(
        "Dữ liệu còn NaN, cần kiểm tra bước C02"
    )

if not np.isfinite(
    X[NUMERIC_FEATURES].to_numpy(dtype=float)
).all():
    raise ValueError("Train có giá trị vô hạn")

if not np.isfinite(
    X_valid[NUMERIC_FEATURES].to_numpy(dtype=float)
).all():
    raise ValueError("Validation có giá trị vô hạn")

LABELS = sorted(y.unique().tolist())

if not set(y_valid.unique()).issubset(set(LABELS)):
    raise ValueError(
        "Validation có nhãn không xuất hiện trong Train"
    )

if y.value_counts().min() < N_SPLITS:
    raise ValueError(
        "Mỗi lớp cần ít nhất 3 mẫu cho 3-fold CV"
    )

# ========================================
# 4. HÀM SIGNED-LOG
# ========================================

def signed_log(values):
    """
    Signed-log:
        sign(x) * log(1 + abs(x))

    Giữ dấu âm/dương và nén độ lớn.
    """

    values = np.asarray(
        values,
        dtype=np.float64
    )

    return (
        np.sign(values)
        * np.log1p(np.abs(values))
    )


# ========================================
# 5. TẠO PIPELINE E12
# ========================================

def make_e12_pipeline():

    # Tương thích các phiên bản sklearn
    ohe_args = {
        "handle_unknown": "ignore"
    }

    if "sparse_output" in inspect.signature(
        OneHotEncoder
    ).parameters:
        ohe_args["sparse_output"] = False
    else:
        ohe_args["sparse"] = False

    # Numeric: Signed-log -> StandardScaler
    numeric_pipeline = Pipeline([
        (
            "signed_log",
            FunctionTransformer(
                signed_log,
                validate=False
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    # Tiền xử lý toàn bộ đặc trưng
    preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                numeric_pipeline,
                NUMERIC_FEATURES
            ),
            (
                "categorical",
                OneHotEncoder(**ohe_args),
                CATEGORICAL_FEATURES
            )
        ],
        remainder="drop",
        sparse_threshold=0
    )

    # MLP giữ nguyên Baseline
    mlp = MLPClassifier(
        hidden_layer_sizes=(64,),
        activation="relu",
        solver="adam",
        alpha=0.0001,
        learning_rate_init=0.001,
        batch_size=512,
        max_iter=60,
        n_iter_no_change=8,
        early_stopping=False,
        random_state=RANDOM_STATE,
        verbose=False
    )

    return Pipeline([
        ("preprocess", preprocessor),
        ("model", mlp)
    ])


# ========================================
# 6. HÀM ĐÁNH GIÁ
# ========================================

METRICS = [
    "accuracy",
    "macro_f1",
    "weighted_f1",
    "macro_precision",
    "macro_recall"
]


def evaluate(y_true, y_pred):

    return {
        "accuracy": accuracy_score(
            y_true, y_pred
        ),

        "macro_f1": f1_score(
            y_true,
            y_pred,
            labels=LABELS,
            average="macro",
            zero_division=0
        ),

        "weighted_f1": f1_score(
            y_true,
            y_pred,
            labels=LABELS,
            average="weighted",
            zero_division=0
        ),

        "macro_precision": precision_score(
            y_true,
            y_pred,
            labels=LABELS,
            average="macro",
            zero_division=0
        ),

        "macro_recall": recall_score(
            y_true,
            y_pred,
            labels=LABELS,
            average="macro",
            zero_division=0
        )
    }


# ========================================
# 7. KIỂM TRA PIPELINE
# ========================================

print("=== CẤU HÌNH E12 ===")

print("Train:", len(X))
print("Validation:", len(X_valid))
print("Số lớp:", len(LABELS))
print("Số đặc trưng số:", len(NUMERIC_FEATURES))
print("Categorical:", CATEGORICAL_FEATURES)

print("\n=== KIỂM TRA SIGNED-LOG ===")

sample_values = np.array(
    [-100, -10, -1, 0, 1, 10, 100],
    dtype=float
)

display(pd.DataFrame({
    "original": sample_values,
    "signed_log": signed_log(sample_values)
}))

print("\n=== PIPELINE E12 ===")
print(make_e12_pipeline())
```

### Xây dựng Pipeline MLP + Signed-log + StandardScaler

**Công thức Signed-log:**

f(x) = sign(x) × ln(1 + |x|)

**Ý nghĩa:**

Signed-log làm giảm độ lớn của các giá trị số,
đồng thời giữ dấu âm/dương.

StandardScaler được áp dụng sau Signed-log
để chuẩn hóa dữ liệu dựa trên trung bình
và độ lệch chuẩn.


**Kết luận:**

Đã xây dựng Pipeline E12 với phép biến đổi
Signed-log trước StandardScaler.

# 3. Huấn luyện E12 bằng 3-ford Cross-Validation


```python
# ========================================
# 3-FOLD CROSS-VALIDATION E12
# ========================================

cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

fold_rows = []
per_class_rows = []
loss_curves = {}

for fold, (fit_idx, cv_idx) in enumerate(
    cv.split(X, y),
    start=1
):

    print("\n" + "=" * 55)
    print(f"E12 - FOLD {fold}/{N_SPLITS}")
    print("=" * 55)

    # Dữ liệu huấn luyện trong fold
    X_fit = X.iloc[fit_idx]
    y_fit = y.iloc[fit_idx]

    # Dữ liệu kiểm định trong fold
    X_cv = X.iloc[cv_idx]
    y_cv = y.iloc[cv_idx]

    # Khởi tạo Pipeline mới cho mỗi fold
    fold_pipeline = make_e12_pipeline()

    start_time = time.perf_counter()

    # Huấn luyện
    fold_pipeline.fit(X_fit, y_fit)

    elapsed = time.perf_counter() - start_time

    # Dự đoán
    pred_cv = fold_pipeline.predict(X_cv)
    pred_train = fold_pipeline.predict(X_fit)

    # Đánh giá
    cv_scores = evaluate(y_cv, pred_cv)
    train_scores = evaluate(y_fit, pred_train)

    trained_mlp = fold_pipeline.named_steps["model"]

    # Lưu Training Loss
    loss_curves[fold] = list(
        trained_mlp.loss_curve_
    )

    # Lưu kết quả từng fold
    fold_rows.append({
        "fold": fold,
        "n_train": len(X_fit),
        "n_cv": len(X_cv),
        "epochs": int(trained_mlp.n_iter_),
        "training_time_sec": float(elapsed),
        "final_train_loss": float(trained_mlp.loss_),

        **cv_scores,

        **{
            f"train_{m}": train_scores[m]
            for m in METRICS
        }
    })

    # Đánh giá theo từng lớp
    p, r, f, support = (
        precision_recall_fscore_support(
            y_cv,
            pred_cv,
            labels=LABELS,
            zero_division=0
        )
    )

    for label, precision, recall, f1_value, n in zip(
        LABELS,
        p,
        r,
        f,
        support
    ):

        per_class_rows.append({
            "fold": fold,
            "class": label,
            "precision": float(precision),
            "recall": float(recall),
            "f1": float(f1_value),
            "support": int(n)
        })

    print("Số mẫu Train:", len(X_fit))
    print("Số mẫu CV:", len(X_cv))
    print("Số epoch:", trained_mlp.n_iter_)
    print(f"Thời gian: {elapsed:.2f} giây")

    print("\n=== KẾT QUẢ FOLD ===")

    for metric in METRICS:
        print(
            f"{metric:18s}: {cv_scores[metric]:.6f}"
        )


# ========================================
# TỔNG HỢP 3 FOLD
# ========================================

cv_results = pd.DataFrame(fold_rows)
cv_per_class = pd.DataFrame(per_class_rows)

cv_summary = pd.DataFrame({
    "metric": METRICS,

    "mean": [
        float(cv_results[m].mean())
        for m in METRICS
    ],

    "std": [
        float(cv_results[m].std(ddof=1))
        for m in METRICS
    ]
})

cv_per_class_summary = (
    cv_per_class
    .groupby("class", as_index=False)
    .agg(
        precision_mean=("precision", "mean"),
        recall_mean=("recall", "mean"),
        f1_mean=("f1", "mean"),
        support_total=("support", "sum")
    )
    .sort_values("f1_mean")
)

print("\n=== KẾT QUẢ 3 FOLD ===")

display(
    cv_results[
        [
            "fold",
            *METRICS,
            "epochs",
            "training_time_sec"
        ]
    ].round(6)
)

print("\n=== TRUNG BÌNH 3 FOLD ===")
display(cv_summary.round(6))

print("\n=== 5 LỚP CÓ F1 THẤP NHẤT ===")

display(
    cv_per_class_summary
    .head(5)
    .round(6)
)

# ========================================
# LƯU CSV VÀO FOLDER E12
# ========================================

cv_results.to_csv(
    RESULT_DIR / "E12_cv_folds.csv",
    index=False,
    encoding="utf-8-sig"
)

cv_summary.to_csv(
    RESULT_DIR / "E12_cv_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

cv_per_class.to_csv(
    RESULT_DIR / "E12_cv_per_class.csv",
    index=False,
    encoding="utf-8-sig"
)

cv_per_class_summary.to_csv(
    RESULT_DIR / "E12_cv_per_class_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nĐã lưu kết quả 3-fold CV.")
```

### Huấn luyện và đánh giá 3-fold Cross-Validation

**Quy trình:**
- Sử dụng StratifiedKFold với 3 fold.
- Mỗi fold tạo một Pipeline E12 mới.
- Huấn luyện trên 2 fold và kiểm định trên 1 fold.
- Signed-log và StandardScaler được thực hiện trong Pipeline.

**Thông tin bổ sung:**
- Số epoch của từng fold.
- Thời gian huấn luyện.
- Training Loss.
- F1-score theo từng lớp.
- Trung bình và độ lệch chuẩn của 3 fold.

# 4. Huấn luyện toàn bộ Train và đánh giá Validation

```python
# ========================================
# HUẤN LUYỆN FULL TRAIN E12
# ========================================

final_model = make_e12_pipeline()

start_time = time.perf_counter()

final_model.fit(X, y)

full_train_time = (
    time.perf_counter() - start_time
)

# Dự đoán Train và Validation
pred_train = final_model.predict(X)
pred_valid = final_model.predict(X_valid)

# Đánh giá
train_scores = evaluate(y, pred_train)
valid_scores = evaluate(y_valid, pred_valid)

final_mlp = final_model.named_steps["model"]

print("=== E12 - FULL TRAIN ===")
print("Số mẫu:", len(X))
print("Số epoch:", final_mlp.n_iter_)
print(f"Thời gian: {full_train_time:.2f} giây")
print(f"Training Loss: {final_mlp.loss_:.6f}")

# ========================================
# BẢNG TRAIN / VALIDATION
# ========================================

validation_metrics = pd.DataFrame({
    "metric": METRICS,

    "train": [
        train_scores[m]
        for m in METRICS
    ],

    "validation": [
        valid_scores[m]
        for m in METRICS
    ]
})

validation_metrics["train_valid_gap"] = (
    validation_metrics["train"]
    - validation_metrics["validation"]
)

print("\n=== TRAIN / VALIDATION ===")

display(validation_metrics.round(6))

# ========================================
# ĐÁNH GIÁ THEO TỪNG LỚP
# ========================================

p, r, f, support = (
    precision_recall_fscore_support(
        y_valid,
        pred_valid,
        labels=LABELS,
        zero_division=0
    )
)

validation_per_class = pd.DataFrame({
    "class": LABELS,
    "precision": p,
    "recall": r,
    "f1": f,
    "support": support
})

print("\n=== KẾT QUẢ TỪNG LỚP ===")

display(
    validation_per_class
    .sort_values("f1")
    .round(6)
)

# ========================================
# CONFUSION MATRIX
# ========================================

cm = confusion_matrix(
    y_valid,
    pred_valid,
    labels=LABELS
)

cm_df = pd.DataFrame(
    cm,
    index=LABELS,
    columns=LABELS
)

# ========================================
# TỔNG HỢP CV / VALIDATION E12
# ========================================

cv_vs_validation = (
    cv_summary[
        ["metric", "mean", "std"]
    ]
    .merge(
        validation_metrics[
            ["metric", "validation"]
        ],
        on="metric"
    )
)

cv_vs_validation["validation_minus_cv_mean"] = (
    cv_vs_validation["validation"]
    - cv_vs_validation["mean"]
)

print("\n=== CV / VALIDATION CỦA E12 ===")

display(cv_vs_validation.round(6))

# ========================================
# LƯU CSV
# ========================================

validation_metrics.to_csv(
    RESULT_DIR / "E12_validation_metrics.csv",
    index=False,
    encoding="utf-8-sig"
)

validation_per_class.to_csv(
    RESULT_DIR / "E12_validation_per_class.csv",
    index=False,
    encoding="utf-8-sig"
)

cm_df.to_csv(
    RESULT_DIR / "E12_validation_confusion_matrix.csv",
    encoding="utf-8-sig"
)

cv_vs_validation.to_csv(
    RESULT_DIR / "E12_cv_vs_validation.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nĐã lưu kết quả Validation.")
```

### Huấn luyện toàn bộ Train và đánh giá Validation

**Mục tiêu:**

Đánh giá khả năng phân loại của mô hình MLP
khi kết hợp Signed-log và StandardScaler.

**Quy trình:**
- Khởi tạo một Pipeline E12 mới.
- Huấn luyện trên toàn bộ tập Train.
- Dự đoán trên tập Validation.
- Ghi nhận 5 chỉ số đánh giá.
- Đánh giá Precision, Recall và F1 theo từng lớp.
